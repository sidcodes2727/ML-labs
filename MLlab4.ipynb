```python
import pandas as pd
df = pd.read_csv("/kaggle/input/mushroom-classification/mushrooms.csv")
print(df.columns)
print(df.duplicated().sum())
print(df.isnull().sum())
for c in df.columns:
    print(c)
    print(df[c].unique())
    print()
df
```

```python
df["class"].value_counts()
```

```python
print(df["stalk-root"].mode())
req = df['stalk-root'].mode()[0]
df['stalk-root'] = df['stalk-root'].replace('?', req)
print(df['stalk-root'].unique())
```

```python
from sklearn.preprocessing import LabelEncoder
enc = LabelEncoder()
for col in df.columns:
    df[col] = enc.fit_transform(df[col])
df
```

```python
from sklearn.model_selection import train_test_split
X= df.drop(columns=['class'])
y= df['class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.naive_bayes import CategoricalNB
model = CategoricalNB()
model.fit(X_train,y_train)
y_pred= model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
def calc(y_train):
  tot = len(y_train)
  pedible=0
  ppoi=0
  for c in y_train:
    if c==0:
      pedible+=1
    else:
      ppoi+=1

  return pedible/tot,ppoi/tot
```

```python
def conditionalProb(X_train,y_train,feature,value,c):
    sub= X_train[y_train==c]
    cnt= sum(sub[feature]==value)
    tot= len(sub)
    k= X_train[feature].nunique()
    p= (cnt+1)/(tot+k)

    return p
```

```python
def predict(inp):
  pedible, ppoi = calc(y_train)
  for fea in X_train.columns:
    val= inp[fea]
    pedible*=conditionalProb(X_train,y_train,fea,val,0)
    ppoi*=conditionalProb(X_train,y_train,fea,val,1)
  if pedible> ppoi:
    return 0

  return 1
```

```python
y_pred=[]
for i,row in X_test.iterrows():
  out= predict(row)
  y_pred.append(out)
```

```python
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
```
