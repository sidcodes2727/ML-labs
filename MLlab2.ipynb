```python
import numpy as np
def entropy(fea):
  # fea.value_counts()
  cnts= fea.value_counts()
  tot= len(fea)
  req= 0;
  for i in cnts:
    p=i/tot
    req=req-p*np.log2(p)

  return req
# entropy(df['class'])
```

```python
def infoGain(data, fea):
  parentEntropy = entropy(data['class'])
  sumchildEntropy = 0
  for i in data[fea].unique():
    req = data[data[fea] == i]
    prop = len(req)/len(data)
    sumchildEntropy+= prop* entropy(req['class'])

  return parentEntropy - sumchildEntropy
```

```python
features= list(df.columns)
features.remove('class')

for feature in features:
    print(feature,":", infoGain(df, feature))
```

```python
def id3(data, features):
    if entropy(data['class'])==0: return data['class'].iloc[0]

    if len(features)==0:  return data['class'].mode()[0]
    gains=[]

    for f in features:
        temp = infoGain(data,f)
        print(f"{f} :", temp)
        gains.append(temp)

    idx = np.argmax(gains)
    bestFeature = features[idx]
    print(f"best feature - ", bestFeature)
    print()
    tree = {bestFeature:{}}

    remFeatures= []
    for f in features:
      if f!=bestFeature:
        remFeatures.append(f)

    for val in data[bestFeature].unique():
        filter= data[data[bestFeature]==val]
        tree[bestFeature][val]= id3(filter,remFeatures)

    return tree
```

```python
def predict(query, tree):

    for node in tree.keys():
        value= query[node]
        res= tree[node][value]

        if isinstance(res,dict):
            return predict(query, res)
        else:
            return res
```

```python
train_df = X_train.copy()
train_df["class"] = y_train
features = list(X_train.columns)
DeciTree = id3(train_df, features)
```

```python
DeciTree
```

```python
y_pred = []

for i, row in X_test.iterrows():
  out= predict(row, DeciTree)
  y_pred.append(out)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
