```python
from sklearn.model_selection import train_test_split
X= df.drop(columns=['class'])
y= df['class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
import numpy as np
def entropy(fea):
  # fea.value_counts()
  cnts= fea.value_counts()
  tot= len(fea)
  req= 0;
  for i in cnts:
    p=i/tot
    req=req-p*np.log2(p)

  return req
# entropy(df['class'])
```

```python
def infoGain(data, fea):
  parentEntropy = entropy(data['class'])
  sumchildEntropy = 0
  for i in data[fea].unique():
    req = data[data[fea] == i]
    prop = len(req)/len(data)
    sumchildEntropy+= prop* entropy(req['class'])

  return parentEntropy - sumchildEntropy
```

```python
def gainRatio(data, fea):
  gain= infoGain(data, fea)
  splitInfo=0
  for i in data[fea].unique():
    req= data[data[fea] == i]
    prop= len(req)/len(data)
    splitInfo= splitInfo -prop*np.log2(prop)

  if splitInfo==0:  return 0

  r = gain/splitInfo
  return r
```

```python
def giniIdx(fea):
  cnts= fea.value_counts()
  tot= len(fea)
  req= 1;
  for i in cnts:
    p=i/tot
    req-=p*p

  return req
```

```python
def giniGain(data,fea):
  parentGini = giniIdx(data['class'])
  sumChildGini= 0
  for i in data[fea].unique():
    req= data[data[fea] == i]
    prop= len(req)/len(data)
    sumChildGini+= prop* giniIdx(req['class'])

  return parentGini-sumChildGini
```

```python
def miscError(fea):
  cnts= fea.value_counts()
  tot= len(fea)
  mx= 0
  for i in cnts:
    mx= max(mx,i)

  return 1-(mx/tot)
```

```python
def miscGain(data, fea):
  parentMisc = miscError(data['class'])
  sumChildMisc= 0
  for i in data[fea].unique():
    req = data[data[fea] == i]
    prop= len(req)/len(data)
    sumChildMisc+=prop* miscError(req['class'])
  return parentMisc- sumChildMisc
```

```python
def id3(data, features, func):
    if entropy(data['class'])==0: return data['class'].iloc[0]

    if len(features)==0:  return data['class'].mode()[0]
    gains=[]

    for f in features:
        temp = func(data,f)
        print(f"{f} :", temp)
        gains.append(temp)

    idx= np.argmax(gains)
    bestFeature= features[idx]
    print(f"best feature - ", bestFeature)
    print()
    tree = {bestFeature:{}}

    remFeatures= []
    for f in features:
      if f!=bestFeature:
        remFeatures.append(f)

    for val in data[bestFeature].unique():
        filter= data[data[bestFeature]==val]
        tree[bestFeature][val]= id3(filter,remFeatures,func)

    return tree
```

```python
def predict(query, tree):

    for node in tree.keys():
        value= query[node]
        res= tree[node][value]

        if isinstance(res,dict):
            return predict(query, res)
        else:
            return res
```

```python
train_df = X_train.copy()
train_df["class"] = y_train
features = list(X_train.columns)
```

```python
print("InfoGain:\n")
dt1= id3(train_df, features, infoGain)
print("gainratior:\n")
dt2= id3(train_df, features, gainRatio)
print("Gini Index:\n")
dt3= id3(train_df, features, giniGain)
print("miscError:\n")
dt4= id3(train_df, features, miscGain)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
def display(DeciTree):
  y_pred = []
  for i, row in X_test.iterrows():
    out= predict(row, DeciTree)
    y_pred.append(out)
  print(classification_report(y_test, y_pred))
  print(accuracy_score(y_test, y_pred))
```
